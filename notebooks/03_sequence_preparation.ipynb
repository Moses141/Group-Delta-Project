# Notebook 3 � Sequence Preparation for LSTM

Prepare supervised sequences from monthly demand:
- Normalize demand using MinMaxScaler
- Sequence length = 12 months
- Target = next month demand
- Save sequences to `outputs/lstm_sequences.npy`


```python
from pathlib import Path
import sys


ROOT = Path("..").resolve()
sys.path.append(str(ROOT))


import numpy as np
import pandas as pd
from app.forecasting import prepare_global_sequences


monthly = pd.read_csv(ROOT / "outputs" / "monthly_demand.csv")
monthly["month"] = pd.to_datetime(monthly["month"])

X, y, _ = prepare_global_sequences(monthly, seq_len=12)

np.save(ROOT / "outputs" / "lstm_sequences.npy", {"X": X, "y": y}, allow_pickle=True)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

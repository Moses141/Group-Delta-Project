# Notebook 4 � LSTM Training

Train one global LSTM model using sequence length 12 and save model artifacts.


```python
from pathlib import Path
import sys


ROOT = Path("..").resolve()
sys.path.append(str(ROOT))


import pandas as pd
from app.forecasting import train_global_lstm

monthly = pd.read_csv(ROOT / "outputs" / "monthly_demand.csv")
monthly["month"] = pd.to_datetime(monthly["month"])

metrics = train_global_lstm(
    monthly_demand=monthly,
    model_path=ROOT / "models" / "lstm_model.h5",
    scaler_path=ROOT / "models" / "lstm_scaler.npy",
    sequence_path=ROOT / "outputs" / "lstm_sequences.npy",
    seq_len=12,
    epochs=30,
    batch_size=16,
)

print(metrics)
```

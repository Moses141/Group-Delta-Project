```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from ml.train import train_demand_pipeline, save_pipeline_artifact
```

```python
df = pd.read_csv('uganda_drug_supply_synthetic.csv')
trained = train_demand_pipeline(
    df=df,
    model=RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1),
    test_size=0.2,
)
trained['metrics']
```

```python
print('Holdout metrics only')
print(trained['metrics'])
print('Train period:', trained['train_period'])
print('Test period:', trained['test_period'])
```

```python
model = trained['pipeline'].named_steps['model']
if hasattr(model, 'feature_importances_'):
    print('Feature importances available from fitted estimator.')
else:
    print('Feature importances unavailable for current estimator.')
```

```python
artifact = save_pipeline_artifact(trained)
print('Saved pipeline artifact:', artifact)
```
